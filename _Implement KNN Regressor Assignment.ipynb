```python
x = titanic.loc[::,titanic.columns.difference(['Survived'])]
y = titanic['Survived']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=10)
```

```python
imputer=SimpleImputer(strategy='most_frequent')
x_train[['Embarked']]=imputer.fit_transform(x_train[["Embarked"]])
x_test[['Embarked']]=imputer.fit_transform(x_test[["Embarked"]])
```

```python
num = x_train.select_dtypes(['int64','float']).columns
Knn_imputer=KNNImputer(n_neighbors=4,weights='uniform')
x_train[num]=Knn_imputer.fit_transform(x_train[num])
x_test[num]=Knn_imputer.transform(x_test[num])
```

```python
x_train=pd.get_dummies(x_train,columns=['Deck','Embarked','Sex'],drop_first=True)
x_test=pd.get_dummies(x_test,columns=['Deck','Embarked','Sex'],drop_first=True)
x_train, x_test = x_train.align(x_test,join='left',axis=1,fill_value=0)
```

```python
#StandardScaler is applied because KNN is a distance-based algorithm. Features with larger scales can dominate the distance calculation.
scale=StandardScaler()
num_col = x_train.select_dtypes(["int64",'float']).columns
x_train[num_col]=scale.fit_transform(x_train[num_col])
x_test[num_col]=scale.transform(x_test[num_col])
```

```python
knn=KNeighborsClassifier(n_neighbors=4,weights="uniform")
```

```python
knn.fit(x_train,y_train)
```

```python
pred_y = knn.predict(x_train)
```

```python
accuracy_score(y_train,pred_y)
```

```python
accuracy_score(y_train,pred_y)
```

```python
pred_y_test = knn.predict(x_test)
```

```python
accuracy_score(y_test,pred_y_test)
```
